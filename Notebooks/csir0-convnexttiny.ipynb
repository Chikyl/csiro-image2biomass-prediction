  <p style="margin-bottom: 15px;">
    The pipeline for predicting pasture biomass is designed to leverage visual patterns effectively while providing accurate multi-target regression:
  </p>

  <ul style="padding-left: 22px; margin-bottom: 0;">
    <li style="margin-bottom: 10px;">Each pasture image is vertically split into left and right halves </li>
    <li style="margin-bottom: 10px;">Each half-image is passed through a pretrained ConvNeXt backbone to extract high-level visual features.</li>
    <li style="margin-bottom: 10px;">Features from the left and right halves are concatenated to create a unified representation of the entire image.</li>
    <li style="margin-bottom: 10px;">The model has three separate regression heads that predict Green matter, GDM and  Total  biomass individually.</li>
    <li style="margin-bottom: 10px;">Dry Dead, is approximated as the difference between predicted Total  biomass and GDM, while Dry Clover is approximated as the difference between predicted GDM and Green</li>
    <li style="margin-bottom: 10px;">Training uses K-Fold cross-validation, with each fold producing its own best model to maximize data utilization and reduce overfitting.</li>
    <li style="margin-bottom: 10px;">On inference, predictions from the five best fold models are averaged to produce final predictions.</li>
    <li style="margin-bottom: 10px;">The model leverages image augmentations (flips, color jitter, etc.) to improve generalization across varying pasture conditions.</li>
    <li style="margin-bottom: 10px;">Performance is evaluated using a weighted R² metric that accounts for the relative importance of each biomass component.</li>
  </ul>

```python
class Config:
    train_path = "/kaggle/input/csiro-biomass/train.csv"
    test_path = "/kaggle/input/csiro-biomass/test"
    sample_sub = "/kaggle/input/csiro-biomass/sample_submission.csv"
    root_dir = "/kaggle/input/csiro-biomass"
    sample_id = "ID1001187975__Dry_Clover_g"

    NUM_EPOCHS = 50
    PATIENCE = 8

    TARGETS = [
        "Dry_Green_g",
        "GDM_g",
        "Dry_Total_g",
        "Dry_Dead_g",
        "Dry_Clover_g",
    ]

    pre_trained_models = "/kaggle/input/my-pretrained-models"



config = Config()


def weighted_r2_score(y_true, y_pred):
    weights = np.array([0.1, 0.1, 0.1, 0.2, 0.5])
    r2s = []

    for i in range(5):
        yt, yp = y_true[:, i], y_pred[:, i]
        ss_res = np.sum((yt - yp) ** 2)
        ss_tot = np.sum((yt - yt.mean()) ** 2)
        r2 = 1 - ss_res / ss_tot if ss_tot > 0 else 0
        r2s.append(r2)

    r2s = np.array(r2s)
    return np.sum(r2s * weights) / weights.sum(), r2s


def calc_metric(outputs, targets):
    y_true = np.column_stack((
        targets[:, 0],
        targets[:, 1],
        targets[:, 2],
        targets[:, 2] - targets[:, 1],
        targets[:, 1] - targets[:, 0],
    ))

    y_pred = np.column_stack((
         outputs[:, 0],
         outputs[:, 1],
         outputs[:, 2],
         outputs[:, 2] - outputs[:, 1],
         outputs[:, 1] - outputs[:, 0],
    ))

    return weighted_r2_score(y_true, y_pred)[0]
```

4. Custom Dataset

  <p style="margin-bottom: 10px;">
    We define a PyTorch <code>Dataset</code> that loads each pasture image, splits it vertically into left and right halves, and returns both views along with the regression targets.
  </p>

  <ul style="padding-left: 22px; margin-bottom: 10px;">
    <li>Images are read from disk and converted to RGB.</li>
    <li>Each image is split into left and right crops to capture spatial variation.</li>
    <li>Both halves are independently transformed and fed to the model.</li>
    <li>The target is a 3-element vector: Green, GDM, and Total.</li>
  </ul>


</div>


```python
class BiomassDataset(Dataset):
    def __init__(self, df, root_dir, transform, train=True):
        self.train = train
        self.root_dir = root_dir
        self.transform = transform

        if self.train:
            self.df = df.reset_index(drop=True)
        else:
            self.items = list(df)

    def __len__(self):
        return len(self.df) if self.train else len(self.items)

    def __getitem__(self, idx):
        if self.train:
            row = self.df.iloc[idx]
            img_path = row["image_path"]
        else:
            img_path, sub_df = self.items[idx]

        img = Image.open(os.path.join(self.root_dir, img_path)).convert("RGB")

        w, h = img.size
        img_left = img.crop((0, 0, w // 2, h))
        img_right = img.crop((w // 2, 0, w, h))

        img_left = self.transform(img_left)
        img_right = self.transform(img_right)

        if self.train:
            target = torch.tensor([
                row['Dry_Green_g'],
                row['GDM_g'],
                row['Dry_Total_g'],
            ], dtype=torch.float32)


            return img_left, img_right, target
        else:
            return img_left, img_right, sub_df


def collate_fn(batch):
    lefts = torch.stack([b[0] for b in batch])
    rights = torch.stack([b[1] for b in batch])
    thirds = [b[2] for b in batch]

    if torch.is_tensor(thirds[0]):
        thirds = torch.stack(thirds)

    return lefts, rights, thirds

```

5. ConvNeXt Regressor Model
</h1>

<div style="
  background: white;
  border-radius: 12px;
  border: 1px solid #e0e0e0;
  font-family: Arial, sans-serif;
  color: #333;
  padding: 18px;
  font-size: 15px;
  line-height: 1.6;
  margin-bottom: 25px;
">

  <p style="margin-bottom: 10px;">
    This model adapts a <strong>ConvNeXt-Tiny</strong> backbone into a dual-view regression network for pasture biomass estimation.
  </p>

  <ul style="padding-left: 22px; margin-bottom: 10px;">
    <li>The classification head is removed and ConvNeXt is used  as a feature extractor.</li>
    <li>Left and right image halves are processed independently through the shared backbone.</li>
    <li>Extracted feature maps are concatenated channel-wise to fuse spatial information.</li>
    <li>Three separate regression heads predict Green, GDM, and Total.</li>
  </ul>


</div>


```python
class ConvNeXtRegressor(nn.Module):
    def __init__(self):
        super().__init__()

        self.backbone = timm.create_model(
            "convnext_tiny",
            pretrained=False,
            features_only=False
        )
        self.backbone.head = nn.Identity()

        self.pool = nn.AdaptiveAvgPool2d(1)

        self.fc = nn.Sequential(
            nn.Linear(768 * 2, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 64),
            nn.ReLU()
        )

        self.head_green = nn.Linear(64, 1)
        self.head_clover = nn.Linear(64, 1)
        self.head_dead = nn.Linear(64, 1)

        self.softplus = nn.Softplus(beta=1.0)

    def forward(self, x_left, x_right):
        f_left = self.backbone.forward_features(x_left)
        f_right = self.backbone.forward_features(x_right)

        f = torch.cat([f_left, f_right], dim=1)
        f = self.pool(f).flatten(1)
        f = self.fc(f)

        green = self.softplus(self.head_green(f))
        clover = self.softplus(self.head_clover(f))
        dead = self.softplus(self.head_dead(f))

        logits = torch.cat([green, dead, clover], dim=1)
        return logits
```

  <p style="margin-bottom: 10px;">
    This block defines the core utilities used to optimize the model, evaluate performance, and control training dynamics.
  </p>

  <ul style="padding-left: 22px; margin-bottom: 10px;">
    <li><strong>Training loop</strong> runs one epoch with mixed-precision to improve speed and reduce memory usage.</li>
    <li><strong>Validation loop</strong> switches to evaluation mode and disables gradients.</li>
    <li>Predictions and targets are collected to compute the official weighted R² metric.</li>
    <li>Both validation loss and metric are returned for monitoring.</li>
    <li><strong>Early stopping</strong> monitors validation R² and halts training when no meaningful improvement is observed.</li>
  </ul>



</div>


```python
def train_one_epoch(model, loader, optimizer, criterion, scaler, device):
    model.train()
    total_loss = 0

    for x_left, x_right, y in loader:
        x_left, x_right, y = x_left.to(device), x_right.to(device), y.to(device)
        optimizer.zero_grad()

        with autocast():
            preds = model(x_left, x_right)
            loss = criterion(preds, y)

        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()

        total_loss += loss.item()

    return total_loss / len(loader.dataset)


def validate(model, loader, criterion, device):
    model.eval()
    total_loss = 0
    preds, trues = [], []

    with torch.no_grad():
        for x_left, x_right, y in loader:
            x_left, x_right, y = x_left.to(device), x_right.to(device), y.to(device)
            p = model(x_left, x_right)

            total_loss += criterion(p, y).item()
            preds.append(p.cpu())
            trues.append(y.cpu())

    preds = torch.cat(preds).numpy()
    trues = torch.cat(trues).numpy()

    wr2 = calc_metric(preds, trues)
    return total_loss / len(loader.dataset), wr2


class EarlyStopping:
    def __init__(self, patience=10, min_delta=1e-4):
        self.patience = patience
        self.min_delta = min_delta
        self.best = None
        self.counter = 0
        self.stop = False

    def step(self, score):
        if self.best is None:
            self.best = score
        elif score < self.best + self.min_delta:
            self.counter += 1
            if self.counter >= self.patience:
                self.stop = True
        else:
            self.best = score
            self.counter = 0
```

  <ul style="padding-left: 22px; margin-bottom: 10px;">
    <li>The best checkpoint from each fold is loaded and set to evaluation mode.</li>
    <li>For each batch, predictions are averaged across all models to form an ensemble.</li>

```python
models = []

test_df = pd.read_csv(os.path.join(config.root_dir, "test.csv"))
group_df = test_df.groupby("image_path")

test_tfms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

test_ds = BiomassDataset(
    group_df, config.root_dir, test_tfms, train=False
)

test_loader = DataLoader(
    test_ds,
    batch_size=16,
    shuffle=False,
    num_workers=2,
    collate_fn=collate_fn
)

model_paths = sorted(
    [p for p in os.listdir(config.pre_trained_models) if p.endswith(".pth")]
)

for p in model_paths:
    m = ConvNeXtRegressor().to(device)
    state = torch.load(
        os.path.join(config.pre_trained_models, p),
        map_location=device
    )
    m.load_state_dict(state)
    m.eval()
    models.append(m)

print(f"✅ Loaded {len(models)} models")

preds = []

with torch.no_grad():
    for imgs_left, imgs_right, df_batch in tqdm(test_loader):
        imgs_left = imgs_left.to(device)
        imgs_right = imgs_right.to(device)

        preds_sum = torch.zeros(
            (imgs_left.size(0), 3), device=device
        )

        for m in models:
            preds_sum += m(imgs_left, imgs_right)

        preds_avg = preds_sum / len(models)
        preds_avg = preds_avg.cpu().numpy()

        preds_5 = np.column_stack((
            preds_avg[:, 0],
            preds_avg[:, 1],
            preds_avg[:, 2],
            preds_avg[:, 2] - preds_avg[:, 1],
            preds_avg[:, 1] - preds_avg[:, 0],
            
        ))

        for i, df_img in enumerate(df_batch):
            row_preds = preds_5[i]
            image_id = df_img["sample_id"].iloc[0].split("__")[0]

            for j, target_name in enumerate(config.TARGETS):
                preds.append({
                    "sample_id": f"{image_id}__{target_name}",
                    "target": float(max(0.0, row_preds[j]))
                })

submission = pd.DataFrame(preds)
submission.to_csv("submission.csv", index=False)

print("🎉 Submission saved:", submission.shape)
print(submission.head())
```
